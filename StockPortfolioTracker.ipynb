# Task 2 - Stock Portfolio Tracker

Calculates the total value of a stock portfolio using a hardcoded price dictionary.

**Skills practised:** dictionaries, input/output, `try/except`, arithmetic, functions, CSV file handling.

---
## Module: `stock_tracker.py`

### Step 1 - Stock prices
A hardcoded dictionary maps each stock symbol to its price, exactly as the task describes.

```python
import csv
from datetime import datetime

STOCK_PRICES = {"AAPL": 180, "TSLA": 250, "GOOGL": 140, "MSFT": 330, "AMZN": 135}
```

### Step 2 - Show the available stocks

```python
def show_available_stocks():
    print("\nAvailable stocks:")
    for symbol, price in STOCK_PRICES.items():
        print("  " + symbol + ": $" + str(price))
```

### Step 3 - Collect the user's holdings
The loop keeps asking until the user types `done`. `try/except` catches non-numeric quantities, and `.get(symbol, 0)` adds up repeated entries of the same stock.

```python
def get_portfolio():
    portfolio = {}
    while True:
        symbol = input("\nStock symbol (or 'done'): ").strip().upper()
        if symbol == "DONE":
            break
        if symbol not in STOCK_PRICES:
            print("Unknown stock. Please choose from the list.")
            continue
        try:
            qty = int(input("Quantity of " + symbol + ": "))
            if qty <= 0:
                raise ValueError
        except ValueError:
            print("Please enter a positive whole number.")
            continue
        portfolio[symbol] = portfolio.get(symbol, 0) + qty
    return portfolio
```

### Step 4 - Calculate the total
This function does no input or printing, which makes it easy to test.

```python
def calculate(portfolio):
    """Return (rows, total). Each row is [symbol, quantity, price, value]."""
    rows = []
    total = 0
    for symbol, qty in portfolio.items():
        price = STOCK_PRICES[symbol]
        value = price * qty
        total += value
        rows.append([symbol, qty, price, value])
    return rows, total
```

### Step 5 - Display and save the result

```python
def print_summary(rows, total):
    print("\n--- Portfolio Summary ---")
    for symbol, qty, price, value in rows:
        print(symbol + ": " + str(qty) + " x $" + str(price) + " = $" + str(value))
    print("Total investment: $" + str(total))


def save_csv(rows, total, filename="portfolio.csv"):
    with open(filename, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["Stock", "Quantity", "Price", "Value"])
        writer.writerows(rows)
        writer.writerow([])
        writer.writerow(["Total", "", "", total])
        writer.writerow(["Generated", datetime.now().strftime("%Y-%m-%d %H:%M")])
    print("Saved to " + filename)
```

### Step 6 - Put it together

```python
def main():
    show_available_stocks()
    portfolio = get_portfolio()
    if not portfolio:
        print("No stocks entered.")
        return
    rows, total = calculate(portfolio)
    print_summary(rows, total)
    save_csv(rows, total)
```

### Try it without typing
10 AAPL and 5 TSLA should give 1800 + 1250 = 3050.

```python
rows, total = calculate({"AAPL": 10, "TSLA": 5})
print_summary(rows, total)
save_csv(rows, total, "portfolio_demo.csv")
```

### Run the interactive version

```python
main()
```
